```python
from pathlib import Path
import json
from src.ingestion.chapter_detector import detect_chapters

INPUT_DIR = Path("dataset/processed/raw_extraction/ISPAD-English-2022")
OUTPUT_DIR = Path("dataset/processed/chapters/ISPAD-English-2022")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

def process_chapter_jsons(input_dir: Path, output_dir: Path):
    for json_file in input_dir.glob("*.json"):
        with open(json_file, "r", encoding="utf-8") as f:
            pages = json.load(f)
        chapters = detect_chapters(pages)
        output_file = output_dir / json_file.name
        with open(output_file, "w", encoding="utf-8") as f:
            json.dump(chapters, f, indent=2, ensure_ascii=False)
        print(f"[DONE] Processed: {json_file} → {output_file}")

process_chapter_jsons(INPUT_DIR, OUTPUT_DIR)
```
