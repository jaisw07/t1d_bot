```python
# -------------------------------
# IMPORTS
# -------------------------------

import json
from pathlib import Path

# -------------------------------
# PATHS
# -------------------------------

INPUT_DIR = Path("dataset/processed/chunks/ISPAD-English-2022")
OUTPUT_DIR = Path("dataset/processed/atoms/ISPAD-English-2022")
CHECKPOINT_DIR = OUTPUT_DIR / "_checkpoints"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

# -------------------------------
# PROCESS ALL FILES
# -------------------------------

files = list(INPUT_DIR.glob("**/*.json"))

# filter out checkpoint files/folders
files = [f for f in files if "checkpoints" not in str(f)]

print(f"Found {len(files)} files")

# -------------------------------
# LOOP
# -------------------------------

for idx, file_path in enumerate(files):
    print(f"\n[{idx+1}/{len(files)}] Processing: {file_path.name}")

    try:
        # Load L2 chunks
        with open(file_path, "r", encoding="utf-8") as f:
            l2_chunks = json.load(f)

        # Output path (same filename)
        output_path = OUTPUT_DIR / file_path.name

        # Per-file checkpoint path for crash-safe resume
        checkpoint_path = CHECKPOINT_DIR / f"{file_path.stem}.checkpoint.json"

        # Run L3 generation with resume support
        generate_l3_for_file(
            l2_chunks,
            output_path=str(output_path),
            checkpoint_path=str(checkpoint_path)
        )

    except Exception as e:
        print(f"[ERROR] Failed on {file_path.name}: {e}")
        continue

print("\n✅ ALL FILES PROCESSED")
```
