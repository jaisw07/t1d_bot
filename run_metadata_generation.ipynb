```python
import os
import json
from pathlib import Path

# -----------------------------
# PATHS
# -----------------------------
INPUT_DIR = Path("dataset/processed/chunks/ISPAD-English-2022")
OUTPUT_DIR = Path("dataset/processed/metadata/ISPAD-English-2022")

# Create output directory if not exists
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# -----------------------------
# GET ALL JSON FILES
# -----------------------------
all_files = [
    f for f in INPUT_DIR.glob("*.json")
    if "checkpoint" not in str(f).lower()
]

print(f"Found {len(all_files)} files")

# -----------------------------
# PROCESS EACH FILE
# -----------------------------
for idx, file_path in enumerate(sorted(all_files)):
    try:
        print(f"\n[{idx+1}/{len(all_files)}] Processing: {file_path.name}")

        # Load L2 chunks
        with open(file_path, "r", encoding="utf-8") as f:
            l2_chunks = json.load(f)

        # Output path (same filename)
        output_path = OUTPUT_DIR / file_path.name
        checkpoint_path = Path(str(output_path) + ".checkpoint.json")

        # Run metadata generation
        generate_metadata_for_file(
            l2_chunks=l2_chunks,
            output_path=str(output_path),
            checkpoint_path=checkpoint_path,
        )

    except Exception as e:
        print(f"[ERROR] Failed on {file_path.name}: {e}")

print("\n✅ DONE: All metadata generated")
```
