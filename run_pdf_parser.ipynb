```python
def run_demo(pdf_path: str, output_path: str):
    print(f"[INFO] Parsing PDF: {pdf_path}")

    pages = extract_pdf_raw(pdf_path)

    print(f"[INFO] Total pages extracted: {len(pages)}")

    Path(output_path).parent.mkdir(parents=True, exist_ok=True)

    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(pages, f, indent=2, ensure_ascii=False)

    print(f"[INFO] Output saved to: {output_path}")
```

## Running for all

```python
import json
from pathlib import Path
from src.ingestion.pdf_extractor import extract_pdf_raw


RAW_ROOT = Path("dataset/raw")
OUT_ROOT = Path("dataset/processed/raw_extraction")


def process_pdf(pdf_path: Path, output_path: Path):
    print(f"[INFO] Processing: {pdf_path}")

    pages = extract_pdf_raw(str(pdf_path))

    output_path.parent.mkdir(parents=True, exist_ok=True)

    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(pages, f, indent=2, ensure_ascii=False)

    print(f"[DONE] Saved: {output_path}")


def process_directory(input_dir: Path, output_dir: Path):
    for pdf_file in input_dir.glob("*.pdf"):
        output_file = output_dir / (pdf_file.stem + ".json")
        process_pdf(pdf_file, output_file)


# Case 1: ISPAE single PDF
ispae_pdf = RAW_ROOT / "ISPAE-Diabetes-Guidelines-2017.pdf"
ispae_out = OUT_ROOT / "ISPAE-Diabetes-Guidelines-2017.json"

process_pdf(ispae_pdf, ispae_out)

# Case 2: ISPAD folder (multiple chapter PDFs)
ispad_dir = RAW_ROOT / "ISPAD-English-2022"
ispad_out_dir = OUT_ROOT / "ISPAD-English-2022"

process_directory(ispad_dir, ispad_out_dir)
```
